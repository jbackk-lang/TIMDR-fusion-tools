# fusion-tools - offline demo

Demo pipeline'u TIMDR + Lambda-tau-rho + Model J na syntetycznym przykladowym sygnale
(`data/w7x_mirnov_example.csv` - NIE sa to prawdziwe dane z W7-X, patrz
`data/example_metadata.json`).

Ten notebook byl wczesniej uszkodzony (pusty/niepoprawny JSON) - zostal odtworzony
od zera i uzywa prawdziwych funkcji bibliotecznych z `timdr/`, `latro/`, `model_j/`.

```python
import os, sys
sys.path.insert(0, os.path.abspath(".."))

from parsers.csv_parser import load_csv
from timdr.timdr_filter import timdr
from timdr.timdr_visualization import plot_timdr
from latro.latro_core import latro
from model_j.model_j_detector import model_j

time, signal = load_csv("../data/w7x_mirnov_example.csv")
print(len(signal), "probek wczytanych")
```

```python
window = 64
reduced = timdr(signal, window=window)
lam, tau, rho = latro(signal)
points = model_j(signal, threshold=2.0)

print("TIMDR:", len(reduced), "probek po redukcji")
print("Lambda-tau-rho:", lam, tau, rho)
print("Model J - wykryte punkty skretu:", list(points))
```

```python
%matplotlib inline
plot_timdr(signal, reduced, window=window, time=time)
```
